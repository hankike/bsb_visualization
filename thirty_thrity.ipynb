```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.patches import Rectangle
```

# Collecting All 30-30 Seasons

A 30-30 season is when a player combines for 30 home runs and 30 stolen bases, the ultimate combination of speed and power.

There have been a total of 81 30-30 seasons in MLB history. Most recently, Pete Crow-Armstrong and Elly De La Cruz in the 2026 season.

```python
df_thirty = pd.read_csv("data/thirty thirty.csv")
```

## The Uniqueness of PCA

Crow-Armstrong did something in 2026 that had only occurred six times before him. He obtained a 40-40 season.

But where does PCA's 2026 season rank among the all time 30-30 season rankings, or even just the 40-40 club?

```python
plt.figure(figsize=(9, 6))

plt.scatter(df_thirty['SB'], df_thirty['HR'])

plt.title(
    'MLB Seasons with 30 Home Runs and 30 Stolen Bases',
    y=1.05,
    fontsize=12,
    fontweight='bold'
)

plt.suptitle(
    'Where does PCA\'s 2026 season rank?',
    y=0.92,
    fontsize=10
)

plt.xlabel('SB')
plt.ylabel('HR', rotation=0)

ax = plt.gca()

ax.xaxis.set_label_coords(0.95, 0.03)
ax.yaxis.set_label_coords(0.02, 0.95)

plt.xticks(range(30, 76, 5))
plt.yticks(range(30, 56, 5))

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Remove grid lines
ax.grid(False)

ax.plot(
    [40, 40],
    [40, 56],
    color='grey',
    alpha=0.75,
    linestyle='--'
)

ax.plot(
    [40, 76],
    [40, 40],
    color='grey',
    alpha=0.75,
    linestyle='--'
)

ax.add_patch(
    Rectangle(
        (40, 40),
        36,
        16,
        facecolor='lightgrey',
        alpha=0.5,
        edgecolor='none'
    )
)

ax.text(38.9, 42.65, 'Canseco 88 & Bonds 96', fontsize=7.5, color='black')
ax.text(45, 41, 'Rodriguez 98', fontsize=9, color='black')
ax.text(40, 46.5, 'Soriano 06', fontsize=9, color='black')

# PCA label with line pointing to the point
ax.annotate(
    'PCA',
    xy=(41, 45),          # actual PCA point
    xytext=(50, 47), # location of PCA text
    fontsize=10,
    color='red',
    fontweight='bold',
    arrowprops=dict(
        arrowstyle='-',
        color='red',
        linewidth=1
    )
)

ax.text(57.5, 53, 'Ohtani 24', fontsize=9, color='black')
ax.text(71, 41.5, 'Acuna Jr. 23', fontsize=9, color='black')
ax.text(31.5, 49.5, 'Walker 97', fontsize=8, color='black')
ax.text(66, 39, '@hank_ike', fontsize=6, color='black', alpha=0.5)

plt.show()
```
